```python
from pathlib import Path
import pandas as pd

CSV = Path("../data/PanTEon_Database_metadata_v1.6.1.csv")

COLS = ["Kingdom", "Phylum", "Class", "Order", "Family",
        "Species", "Source", "Source 2", "Notes"]

# Leer todo como texto y sin convertir vacíos a NaN
crudo = pd.read_csv(CSV, dtype=str, keep_default_na=False)

# Comprobar que no falta ninguna columna
faltan = [c for c in COLS if c not in crudo.columns]
assert not faltan, f"Columnas no encontradas: {faltan}"

# Quedarse solo con las 9 columnas, en el orden del enunciado
datos = crudo[COLS].copy()

print(datos.shape)
print(list(datos.columns) == COLS)
datos.head()
```

```python


# quitar espacios en las 9 columnas
tras_strip = datos.apply(lambda col: col.str.strip())


cambios_strip = (datos != tras_strip).sum()
print("Valores modificados por strip:")
print(cambios_strip[cambios_strip > 0] if cambios_strip.sum() else "ninguno")

# vacio a nulo, y forzar None real (no NaN) para que MongoDB guarde null
datos = tras_strip.replace("", None)
datos = datos.astype(object).where(datos.notna(), None)

# Comprobaciones
print("\nNulos por columna:")
print(datos.isna().sum())
print("\nCadenas vacías restantes:", (datos == "").sum().sum())

vacio = datos.loc[datos["Phylum"].isna(), "Phylum"].iloc[0]
print("Tipo de un vacío:", type(vacio).__name__)
```
